```python
# ============================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================

import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# ============================================
# 2. LOAD IMAGE
# ============================================

image_path = "i.jpg"

image = cv2.imread(image_path)

if image is None:
    print("ERROR: Image not found!")
else:
    print("Image loaded successfully!")
    print("Image shape:", image.shape)
```

```python
# ============================================
# 3. CONVERT IMAGE TO GRAYSCALE
# ============================================

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Grayscale image created successfully!")
print("Grayscale shape:", gray.shape)
```

```python
# ============================================
# 4. DISPLAY ORIGINAL IMAGE
# ============================================

plt.figure(figsize=(6, 5))

plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.show()
```

```python
# ============================================
# 5. ORIGINAL IMAGE HISTOGRAM
# ============================================

plt.figure(figsize=(8, 5))

plt.hist(gray.ravel(), bins=256, range=[0, 256])

plt.title("Original Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

```python
# ============================================
# 6. CONTRAST STRETCHING
# ============================================

contrast_stretched = cv2.normalize(
    gray,
    None,
    0,
    255,
    cv2.NORM_MINMAX
)

print("Contrast stretching completed!")
```

```python
# ============================================
# 7. DISPLAY CONTRAST STRETCHED IMAGE
# ============================================

plt.figure(figsize=(6, 5))

plt.imshow(contrast_stretched, cmap="gray")
plt.title("Contrast Stretched Image")
plt.axis("off")

plt.show()
```

```python
# ============================================
# 8. HISTOGRAM EQUALIZATION
# ============================================

hist_equalized = cv2.equalizeHist(gray)

print("Histogram Equalization completed!")
```

```python
# ============================================
# 9. DISPLAY HISTOGRAM EQUALIZED IMAGE
# ============================================

plt.figure(figsize=(6, 5))

plt.imshow(hist_equalized, cmap="gray")
plt.title("Histogram Equalized Image")
plt.axis("off")

plt.show()
```

```python
# ============================================
# 10. CLAHE
# ============================================

clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8, 8)
)

clahe_image = clahe.apply(gray)

print("CLAHE enhancement completed!")
```

```python
# ============================================
# 11. DISPLAY CLAHE IMAGE
# ============================================

plt.figure(figsize=(6, 5))

plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE Enhanced Image")
plt.axis("off")

plt.show()
```

```python
# ============================================
# 12. COMPARE ALL ENHANCED IMAGES
# ============================================

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(contrast_stretched, cmap="gray")
plt.title("Contrast Stretching")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(hist_equalized, cmap="gray")
plt.title("Histogram Equalization")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 13. COMPARE HISTOGRAMS
# ============================================

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.hist(gray.ravel(), bins=256, range=[0, 256])
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 2)
plt.hist(
    contrast_stretched.ravel(),
    bins=256,
    range=[0, 256]
)
plt.title("Contrast Stretching Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 3)
plt.hist(
    hist_equalized.ravel(),
    bins=256,
    range=[0, 256]
)
plt.title("Histogram Equalized")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 4)
plt.hist(
    clahe_image.ravel(),
    bins=256,
    range=[0, 256]
)
plt.title("CLAHE Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 14. SAVE ENHANCED IMAGES
# ============================================

cv2.imwrite("contrast_stretched.jpg", contrast_stretched)
cv2.imwrite("histogram_equalized.jpg", hist_equalized)
cv2.imwrite("clahe_image.jpg", clahe_image)

print("Images saved successfully!")
print("1. contrast_stretched.jpg")
print("2. histogram_equalized.jpg")
print("3. clahe_image.jpg")
```

```python
# ============================================
# 15. FINAL COMPARISON
# ============================================

plt.figure(figsize=(16, 5))

plt.subplot(1, 4, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(contrast_stretched, cmap="gray")
plt.title("Contrast Stretching")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(hist_equalized, cmap="gray")
plt.title("Histogram Equalization")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE")
plt.axis("off")

plt.tight_layout()
plt.show()
```
