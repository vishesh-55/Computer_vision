```python
# ============================================================
# Experiment No. 3
# Implementation and Analysis of Spatial Filtering Techniques
# using Low-Pass and High-Pass Filters
# ============================================================

import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# ============================================================
# 2. LOAD IMAGE
# ============================================================

image_path = "immm.jpg"

image = cv2.imread(image_path)

if image is None:
    print("ERROR: Image not found!")
else:
    print("Image loaded successfully!")
    print("Image shape:", image.shape)
```

```python
# ============================================================
# 3. CONVERT IMAGE TO GRAYSCALE
# ============================================================

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Grayscale image created successfully!")
print("Grayscale shape:", gray.shape)
```

```python
# ============================================================
# 4. DISPLAY ORIGINAL IMAGE
# ============================================================

plt.figure(figsize=(6, 5))
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# 5. GAUSSIAN FILTERING
# ============================================================

gaussian = cv2.GaussianBlur(gray, (5, 5), 0)

print("Gaussian filtering completed!")
```

```python
# ============================================================
# 6. DISPLAY GAUSSIAN FILTERED IMAGE
# ============================================================

plt.figure(figsize=(6, 5))
plt.imshow(gaussian, cmap="gray")
plt.title("Gaussian Filtered Image")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# 7. MEDIAN FILTERING
# ============================================================

median = cv2.medianBlur(gray, 5)

print("Median filtering completed!")
```

```python
# ============================================================
# 8. DISPLAY MEDIAN FILTERED IMAGE
# ============================================================

plt.figure(figsize=(6, 5))
plt.imshow(median, cmap="gray")
plt.title("Median Filtered Image")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# 9. AVERAGE / MEAN FILTERING
# ============================================================

average = cv2.blur(gray, (5, 5))

print("Average filtering completed!")
```

```python
# ============================================================
# 10. DISPLAY AVERAGE FILTERED IMAGE
# ============================================================

plt.figure(figsize=(6, 5))
plt.imshow(average, cmap="gray")
plt.title("Average Filtered Image")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# 11. LAPLACIAN FILTERING
# ============================================================

laplacian = cv2.Laplacian(gray, cv2.CV_64F)

laplacian = cv2.convertScaleAbs(laplacian)

print("Laplacian filtering completed!")
```

```python
# ============================================================
# 12. DISPLAY LAPLACIAN IMAGE
# ============================================================

plt.figure(figsize=(6, 5))
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Filtered Image")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# 13. SOBEL X FILTERING
# ============================================================

sobel_x = cv2.Sobel(
    gray,
    cv2.CV_64F,
    1,
    0,
    ksize=3
)

sobel_x = cv2.convertScaleAbs(sobel_x)

print("Sobel X filtering completed!")
```

```python
# ============================================================
# 14. DISPLAY SOBEL X IMAGE
# ============================================================

plt.figure(figsize=(6, 5))
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X - Horizontal Edges")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# 15. SOBEL Y FILTERING
# ============================================================

sobel_y = cv2.Sobel(
    gray,
    cv2.CV_64F,
    0,
    1,
    ksize=3
)

sobel_y = cv2.convertScaleAbs(sobel_y)

print("Sobel Y filtering completed!")
```

```python
# ============================================================
# 16. DISPLAY SOBEL Y IMAGE
# ============================================================

plt.figure(figsize=(6, 5))
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y - Vertical Edges")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# 17. COMPARE LOW-PASS FILTERS
# ============================================================

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gaussian, cmap="gray")
plt.title("Gaussian Filter")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(median, cmap="gray")
plt.title("Median Filter")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(average, cmap="gray")
plt.title("Average Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 18. COMPARE HIGH-PASS FILTERS
# ============================================================

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 19. FINAL COMPARISON OF ALL FILTERS
# ============================================================

plt.figure(figsize=(18, 10))

plt.subplot(2, 4, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(gaussian, cmap="gray")
plt.title("Gaussian")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(median, cmap="gray")
plt.title("Median")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(average, cmap="gray")
plt.title("Average")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 20. SAVE FILTERED IMAGES
# ============================================================

cv2.imwrite("gaussian.jpg", gaussian)
cv2.imwrite("median.jpg", median)
cv2.imwrite("average.jpg", average)
cv2.imwrite("laplacian.jpg", laplacian)
cv2.imwrite("sobel_x.jpg", sobel_x)
cv2.imwrite("sobel_y.jpg", sobel_y)

print("Images saved successfully!")
print("1. gaussian.jpg")
print("2. median.jpg")
print("3. average.jpg")
print("4. laplacian.jpg")
print("5. sobel_x.jpg")
print("6. sobel_y.jpg")
```
